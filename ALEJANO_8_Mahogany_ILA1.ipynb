```python
#Fuel Efficiency Calculator

def calculate_fuel_efficiency(distance, fuel_consumed):
    """Calculate fuel efficiency in kilometers per liter."""
    if fuel_consumed <= 0:
        return None

    return distance / fuel_consumed


#Get input from the user
distance = float(input("Enter distance traveled (in kilometers): "))
fuel = float(input("Enter fuel consumed (in liters): "))

#Calculate and display fuel efficiency
efficiency = calculate_fuel_efficiency(distance, fuel)

if efficiency is None:
    print("Fuel consumed must be greater than zero.")
else:
    print(f"Fuel Efficiency: {efficiency} km/L")
```
